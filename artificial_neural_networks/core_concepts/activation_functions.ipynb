## **1. Identity (Linear) Function:**
**Mathematical Formulation:**
 $$theta(z) = z$$

**Properties & Usage:** This function is a linear activation function and used in the output layer of the networks that we try to predict linear continous values or we can say we want to train a linear regression model in neural networks.

```python
import seaborn as sns
import numpy as np

sns.set_theme(context='notebook', style='whitegrid')

def identity(z):
	return z

x = np.linspace(-10, 10, 100)
y = identity(x)

ax = sns.lineplot(x=x, y=y)
ax.set(
	title="Identity Function",
	ylabel="Output",
	xlabel="x"
)
```
