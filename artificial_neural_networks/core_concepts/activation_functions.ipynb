## **2. Unit Step (Heaviside) Function:**

This function is the base activation function in a Perceptron.

**Mathematical Formulation:**

$$
	\text{Unit Step}(z) =
	\begin{cases}
		0 & \text{if } z < 0 \\
		0.5 & \text{if } z = 0 \\
		1 & \text{if } z > 0
	\end{cases}
$$

**Properties & Usage:** This function were used in the early Perceptrons and results in a binary kind output (0 or 1) but it's not the function that used in the output layer of a binary classification and in that case we use Sigmoid function that results in a propability like output and using a suitable threshold we classify the binary target variable.


```python
import numpy as np
import seaborn as sns

sns.set_theme(context='notebook', style='whitegrid')

x = np.linspace(-5, 5, 1000)
y = []
for num in x:
    if num < 0:
        y.append(0)
    elif num == 0:
        y.append(0.5)
    else:
        y.append(1)

sns.lineplot(
    x=np.linspace(-10, 10, 1000), y=np.zeros(shape=(1000, )),
    color='black', alpha=0.7, linestyle='--'
    )

ax = sns.lineplot(x=x, y=y, color='blue')
ax.set(
    title="Unit Step",
    ylabel="y",
    xlabel="x",
    xlim=(-10, 10),
    ylim=(-1, 2)
)
```

## 2. Sign Function:

This function uses when you want `-1` for the negative input values and `+1` for the positive input values. not a non-linear activation function.

**Mathematical Formulation:**
$$
	\text{Sign}(z) =
	\begin{cases}
		-1 & \text{if } z < 0 \\
		0 & \text{if } z = 0 \\
		1 & \text{if } z > 0
	\end{cases}
$$

**Properties & Usage:** This activation function linear and we don't use it any more it was used in early Perceptron as activation function.

```python
import numpy as np
import seaborn as sns

x = np.linspace(-8, 8, 1000)
y = []
for num in x:
    if num < 0:
        y.append(-1)
    elif num == 0:
        y.append(0)
    else:
        y.append(1)

sns.lineplot(
    x=np.linspace(-10, 10, 1000), y=np.zeros(shape=(1000, )),
    color='black', alpha=0.7, linestyle='--'
    )

ax = sns.lineplot(x=x, y=y, color='blue')
ax.set(
    title="Sign Function",
    ylabel="y",
    xlabel="x",
    xlim=(-10, 10),
    ylim=(-2, 2)
)
```

## **3. Piece-Wise Linear Function:**

This function is linear in a specific input range and make the transition from 0 to 1 a little smoother.

**Mathematical Formulation:**

constant value is equal to 0.5 in this example.
$$
	\text{Sign}(z) =
	\begin{cases}
		0 & \text{if } z \le -0.5 \\
		z + 0.5 & \text{if } -0.5 < z < 0.5 \\
		1 & \text{if } z \ge 0.5
	\end{cases}
$$

**Properties & Usage:** This function used in the support vector machines. Our focus in not just on classifying the lables but to find classes that have the most separation from each other. We seek for boudaries that not just classify the lables but to make the most possible separation between the data points from different classes so maybe we will generalize better to new unseen data.

```python
import numpy as np
import seaborn as sns

x = np.linspace(-5, 5, 10000)
y = []
for num in x:
    if num <= -0.5:
        y.append(0)
    elif -0.5 < num < 0.5:
        y.append(num + 0.5)
    elif num >= 0.5:
        y.append(1)

sns.lineplot(
    x=np.linspace(-10, 10, 1000), y=np.zeros(shape=(1000, )),
    color='black', alpha=0.7, linestyle='--'
    )

ax = sns.lineplot(x=x, y=y, color='blue')
ax.set(
    title="Piece-Wise Linear Function",
    ylabel="y",
    xlabel="x",
    xlim=(-8, 8),
    ylim=(-1, 2)
)
```

## **4. Identity (Linear) Function:**

**Mathematical Formulation:**
$$theta(z) = z$$

**Properties & Usage:** This function is a linear activation function and used in the output layer of the networks that we try to predict linear continous values or we can say we want to train a linear regression model in neural networks. The reason that we use this activation function is that this function will return exactly what you input to it that is the thing we need in the regression task.

```python
import seaborn as sns
import numpy as np

def identity(z):
	return z

x = np.linspace(-8, 8, 100)
y = identity(x)

sns.lineplot(
    x=np.linspace(-10, 10, 1000), y=np.zeros(shape=(1000, )),
    color='black', alpha=0.7, linestyle='--'
    )

ax = sns.lineplot(x=x, y=y)
ax.set(
	title="Identity Function",
	ylabel="y",
	xlabel="x",
	xlim=(-10, 10),
	ylim=(-10, 10)
)
```

**Mathematical Formulation:**
$$
	\text{Sigmoid}(z) = \frac{1}{1 + e^{-z}}
$$
