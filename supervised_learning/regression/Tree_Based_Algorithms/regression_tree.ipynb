# Regression Tree

When we're doing a regression task and we enter a data that there is no clear linear pattern between the features and the target variable using a linear regression model will result in a model when poor performance surely cause the relationship that we trying to model is not linear.

Maybe we think so we can use the expansion of linear regression an use polynomial features, this could work when there is a clear non-linear relationship so we can capture the pattern. But polynomial regression hardly suffers from the overfitting problem and needs a lot of hyperparameter tuning and regularization to generalize well to the unseen data.

Another option to capture complex patterns in a regression task is we can use regression trees. There are many similarities between **Decision Trees** and **Regression Trees**:
- **Nodes**
- **Top-Down Induction Process**
- **Recursive Partitioning**
- **Splitting Criterion**
- **Stopping Criteria**

One difference is that the target variable in the regression tree is a continuous random variable and we should predict an exact number. So to predict a number we use the mean of the continuous outputs in the threshlod for the prediction.x

## Splitting Creterion

Another difference is that the splitting creterion here is the **SSR** (Sum of Squared Residuals).

$$
	SSR = (y - \hat y) \\
	\text{ } \\
	\hat y = \text{avg}(\text{values in the threshold})
$$
for a continuous feature:
- We sort the feature ascending, then calculate the mean for each two pairs of data.
- Then we use each of them as a threshold to split the target variable using them.
- We calculate the average of the target variable in each threshold.
- Then for each threshold we calculate the SSR.
- We pick the threshold with least SSR (keeping its SSR for the next step) as the node so we can decrease the impurity most using it.

for a categorical features:
- First split the data using existing categories.
- Then calculate the average of values in each threshold.
- We calculate the SSR.

Now we know how to split the data using numerical or categorical features so we pick the feature with the lowest SSR in each step of the recursive partitioning so using it we can decrease the impurity the most at every step.

## Stopping Criteria

In the regression tree there are bunch of stopping criteria like decision tree that using them we can prevent overfitting and have a logic to stop the training process at some point. Stopping Criteria:
- **Maximum Depth**
- **Minimum Samples** --> common: `20`
- **Maximum SSR**

## Pruning a Regression Tree

Pruning is a regularization technique in the decision trees and regression trees to:
- Stabilize the Model Performace.
- Improve the Predictive Performance on the unseen data.

We have different prunig techniques in the tree-based algorithms:
- Pre-Pruning
- Post-Pruning

Cost-Complexity Pruning is one of the most used pruing techniques for tree-based algorithms and this approach implemented in scikit-learn in **ccp_alpha** parameter and the effective alphas can get from the **cost_complexity_pruning_path** function.
