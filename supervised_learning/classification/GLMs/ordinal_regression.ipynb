# Ordinal Regression:

There are some times in the real-world scenarios that predicting an actual value for the target variable is not realy needed so we convert our target variable to bins (for example: low, medium, high) and now we can deal we a classification task rather than the regression task that make the prediction easier. By converting a regression problem to a classification problem we can make a model that should predict less detailed target variable and can improve the prediction task.

By knowing this case we can say that ordinal target variable is a very common target variable that we need to predict, So this regression model makes us improve prediction on ordinal target variables and use its more accurate and interpretable predictions than the multiclass classification using Logistic regression.

> **Note:** In the probit regression unlike multiclass classification, we don't predict the probability of a class in the target variable, but we predict the probability of a class and other lower classes. We can say we predict the proabilities that a sample refers to a class and the lower classes.

> **Note:** We can say the coefficients of the Probit model show that the features increase the likelihood of a being in a higher class, instead of logistic regression coefficients that they interpret as the increase of log-odds of the sample to be in this class. We can say for a feature a 5 classes the Probit model results in 5 probabilities (sum to 100 percent) and the Logistic regression model results in 5 coefficients that shows the increase or decrease of a prediction to be related to each sample. We see that the interpretaion of Probit model gives us a big picture of the features.

**Main Components:**
- Random Component: `Binomial Distribution`
- Systematic Component: `Linear Predictor`
- Link Function: `Logit`, `Probit`, `Log-Log`, ...

## Mathematical Formulation:

**Linear Predictor:**

$$
	\text{Linear Predictor}: a_j - (\beta_0 + \beta_1 x_1 + \beta_2 x_2 + ... + \beta_k x_k) \\
	\text{or} \\
	\text{Linear Predictor}: a_j - \beta X
$$
- Linear Predictor is the $z$ you see in the texts.
- $a_j$ is the threshold for the $j^{th}$ category.

**Link Functions:**

Logit Link Function:
$$
	\text{Link Function}: \text{Logit}(p) = \log \left( \frac{p}{1 - p} \right) = z \\
	\text{ } \\
	\text{Probability}: p = \frac{1}{1 + e^{-z}}
$$

Probit Link Function:
$$
	\text{Link Function}: \text{Probit}(p) = \Phi^{-1}(p) = z \\
	\text{ } \\
	\text{Probability}: p = \Phi(z)
$$

Log-Log Link Function:
$$
	\text{Link Function}: \text{Log-Log}(p) = -\ln (-\ln p) = z \\
	\text{ } \\
	\text{Probability}: p = \exp (-\exp(-z))
$$

> **Note:** Probit link fucntion is used when you believe that the underlying distribution of independent variables are normally distributed. 

## Advantages & Disadvantages:

**Advantages:**
- respect the order in the target variable
- more interpretable for ordered categories

**Disadvantages:**
- proportional log-odds assumption: assumes that independent variables have similar effect on each category. if this assumption violated we can use multiclass classification or use other models like random forest, ...
- May not be suitable when the ordering in the target variable doesn't makes sense

## Assumptions:

- The effect of independent variables are similar for each category
- Independence of observations
- Ordinal Target variable

```python
# ordinal regression synthetic data

# In[1]:


import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from statsmodels.miscmodels.ordinal_model import OrderedModel
from sklearn.metrics import accuracy_score


# In[2]:


np.random.seed(0)


# In[3]:


def generate_data(n_samples=2_000, n_features=2, n_classes=3, noise=0.1):
    X = np.random.normal(0, 1, size=(n_samples, n_features))
    y = np.random.choice(n_classes, size=n_samples, replace=True)
    
    # add noise to the data
    X += np.random.normal(0, noise, size=(n_samples, n_features))
    
    return X, y


# In[4]:


X, y = generate_data(n_samples=100, n_features=2, n_classes=3, noise=0.1)


# In[5]:


X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
X_train, X_val, y_train, y_val = train_test_split(X_train, y_train, test_size=0.2, random_state=0)


# In[6]:

model = OrderedModel(endog=y_train, exog=X_train)
model_fit = model.fit()
model_fit.summary()

# evaluation
y_pred = model_fit.predict(X_val)
labels = []
for i in range(y_pred.shape[0]):
    label = 0
    for j in range(3):
        if y_pred[i][j] == y_pred.max(axis=1)[i]:
            labels.append(label)
        else:
            label += 1

accuracy_score(y_val, labels)
```
