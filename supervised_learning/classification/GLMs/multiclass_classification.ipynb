# Multiclass Classification

We now want to classify a target variable that is not binary and has several classes. There are two main approaches to do the multiclass classification using Logistic Regression:

- OVR (One Vs. Rest)
- Softmax

> **OVR:** In this approach we train Logistic Regression for each class versus other classes so we can train binary classification models to the number of classes. This approach is very easy to interpret, efficient, and intuitive so we can classify each class from others. At last we choose the class with the highest confidence interval as the label for the prediction.

<img src="../../images/one-vs-rest.jpg" height="400">

> **Softmax:** Softmax is a function that used instead of sigmoid to calculate the probability of the related class using the *z*. This function relates classes to each other so the sum of all the probabilities for each class will result in **1**. Using this method we will get smoother and better decision boundaries and mostly result in better classification model, but it's not interpretable as **OVR**
approach and can be more computationaly expensive.

$$\text{softmax}(z_i) = \frac{e^{z_i}}{\sum_{j=1}^K e^{z_j}}$$
<img src="../../images/softmax.png" height="400">

Softmax Approach is a very similar way of training a neural network model that includes softmax activation functions as output layer.


## Cons & Pros:

**OVR:**

Pros:
- Easily Interpretable
- More Computationally efficient

Cons:
- Does not consider all the classes attributes in every classification

**Softmax:**

Pros:
- Can Capture more nuanced patterns in the data
- Sum of the probability outputs of each prediction equals to 1

Cons:
- Can be more computationally expensive
- Less Interpretable
